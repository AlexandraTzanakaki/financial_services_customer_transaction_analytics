### 4.2 Missing Value Percentages

Missing-value counts are converted into percentages to understand the scale of missingness in each table.

```python
missing_summary = []

for name, df in tables.items():
    total_rows = len(df)
    missing_count = df.isna().sum().sum()
    missing_percentage = (missing_count / df.size) * 100
    
    missing_summary.append({
        "table": name,
        "rows": total_rows,
        "columns": len(df.columns),
        "missing_cells": missing_count,
        "missing_percentage": missing_percentage
    })

missing_summary = pd.DataFrame(missing_summary)

missing_summary
```

# 8. Validate Business Rules

Business-rule checks are performed to identify relationships that are technically valid but require careful interpretation during analysis.

### 8.1 Customers per Account

The number of customers associated with each account is calculated to identify individual and shared-account relationships.

```python
customers_per_account = (
    disp.groupby("account_id")["client_id"]
    .nunique()
)

print(customers_per_account.value_counts().sort_index())
```

### 8.2 Validate Account Ownership

The number of owners associated with each account is checked to identify accounts without an owner or with multiple owners.

```python
owners_per_account = (
    disp[disp["type"] == "OWNER"]
    .groupby("account_id")["client_id"]
    .nunique()
)

print(owners_per_account.value_counts().sort_index())

print("\nAccounts with no owner:")
print((~account["account_id"].isin(owners_per_account.index)).sum())

print("\nAccounts with multiple owners:")
print((owners_per_account > 1).sum())
```

### 8.3 Validate Shared Account Dispositions

Shared accounts are examined to confirm how owner and additional customer relationships are represented in the disposition table.

```python
shared_disp = disp[
    disp["account_id"].isin(
        disp.groupby("account_id")["client_id"].nunique()
        .loc[lambda x: x > 1]
        .index
    )
].copy()
```

```python
shared_account_types = (
    shared_disp
    .groupby(["account_id", "type"])["client_id"]
    .nunique()
    .unstack(fill_value=0)
)

print(shared_account_types.value_counts())

print("\nShared accounts with exactly one OWNER and one DISPONENT:")
print(
    (
        (shared_account_types["OWNER"] == 1) &
        (shared_account_types["DISPONENT"] == 1)
    ).sum()
)

print("\nTotal shared accounts:")
print(len(shared_account_types))
```

### 8.4 Validate Customer-to-Account Structure

The number of unique accounts associated with each client is checked to confirm the customer-account relationship structure.

```python
accounts_per_client = (
    disp.groupby("client_id")["account_id"]
    .nunique()
)

print(accounts_per_client.value_counts().sort_index())

print("\nClients with no account:")
print((~client["client_id"].isin(accounts_per_client.index)).sum())

print("\nClients with multiple accounts:")
print((accounts_per_client > 1).sum())
```

### 11.4 Distribution of the Most Common Transaction Amount Over Time

```python
amount_14_60_by_year = (
    amount_14_60
    .groupby(amount_14_60["transaction_date"].dt.year)
    .size()
    .reset_index(name="transaction_count")
)

amount_14_60_by_year
```

### 11.5 Relative Frequency of the Most Common Transaction Amount

The annual share of transactions with the most common amount is calculated to determine whether its prevalence changed over time independently of the overall transaction volume.

```python
amount_14_60_by_year = (
    amount_14_60
    .groupby(amount_14_60["transaction_date"].dt.year)
    .size()
    .reset_index(name="amount_14_60_count")
)

total_transactions_by_year = (
    trans
    .groupby(trans["transaction_date"].dt.year)
    .size()
    .reset_index(name="total_transaction_count")
)

amount_14_60_by_year = amount_14_60_by_year.merge(
    total_transactions_by_year,
    left_on="transaction_date",
    right_on="transaction_date"
)

amount_14_60_by_year["percentage_of_transactions"] = (
    amount_14_60_by_year["amount_14_60_count"]
    / amount_14_60_by_year["total_transaction_count"]
    * 100
)

amount_14_60_by_year
```
